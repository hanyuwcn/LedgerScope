> - 行业基准来源：WordStream/LocaliQ 2025、Industry Week


### 3.2 规模类指标定义

#### 月线索量（Monthly Leads）

**定义**：每月通过 Google Search 广告产生的询盘数量。这是广告漏斗的中游产出指标，位于点击（上游）和订单（下游）之间。

**计算公式**：
```
Traffic = (Budget × Allocation) / (CPC × USDToRMB)
Leads = Traffic × ConversionRate
```

**业务含义**：反映广告投放的获客规模。线索量不足时，应优先优化广告参数（CPC、CVR）而非销售团队。

**依赖参数**：广告预算、分配比例、CPC、汇率、CVR


#### 月订单量（Monthly Orders）

**定义**：每月成交的订单数量。这是销售漏斗的最终产出指标。

**计算公式**：
```
Orders = Leads × CloseRate
```

**业务含义**：反映销售团队将线索转化为订单的能力。订单量是收入和利润的基础。

**依赖参数**：月线索量、Close Rate


#### 月收入（Monthly Revenue）

**定义**：每月销售总收入，以人民币计价。

**计算公式**：
```
Revenue = UnitFobPriceInRMB × UnitsSold
UnitsSold = Orders × UnitsPerOrder
```

**业务含义**：企业的现金流入口，是所有成本和利润的计算基础。

**依赖参数**：FOB 出口价（人民币）、订单量、每订单套数


### 3.3 利润类指标定义

#### 月毛利（Monthly Gross Profit）

**定义**：月收入减去直接产品成本（COGS）后的余额。

**计算公式**：
```
COGS = UnitExWorksPrice × UnitsSold
GrossProfit = Revenue - COGS
```

**业务含义**：反映产品层面的盈利能力，尚未扣除广告费和运营费用。

**依赖参数**：月收入、采购单价（EXW）、销售套数


#### 单位毛利（Unit Gross Profit）

**定义**：每销售一套产品所带来的毛利贡献。这是盈亏平衡分析的核心输入。

**计算公式**：
```
UnitGrossProfit = GrossProfit / UnitsSold
```

**业务含义**：每套产品在扣除直接采购成本后，可用于覆盖广告费和固定成本的金额。

> **术语说明**：单位毛利即为旧版所称“单位边际贡献”（Unit Contribution Margin），本报告统一使用“单位毛利”。

**依赖参数**：月毛利、销售套数


#### 月净利润（Monthly Net Income）

**定义**：月收入减去所有成本、费用和税金后的最终盈利。

**计算公式**：
```
MonthlyExpense = RentExpense + RenderExpense + TravelExpense
ManagementExpense = MonthlyManagementExpense × Months（默认 Months=1）
AdvertisingExpense = MarketingExpense × 1.0（默认全部用于广告）
BrandFreightExpense = 0.0（运费由下游买方承担）
TotalSellingExpense = FreightExpense + MarketingExpense
TotalExpense = ManagementExpense + SellingExpense

OperatingIncome = Revenue - COGS - TotalExpense - Depreciation（默认 Depreciation=0）
NetIncome = OperatingIncome × (1 - TaxRate)
```

> **说明**：
> - 当前模型假设折旧为 0
> - 物流费用由下游买方承担（Merchant 承担运费），本模型不列支

**业务含义**：企业的最终盈利指标，是衡量商业模式可行性的核心依据。

**依赖参数**：月收入、COGS、月运营费用（租金/差旅/第三方服务费）、广告费、税率


#### 月自由现金流（Monthly Free Cash Flow）

**定义**：月净利润经非现金支出（折旧）和资本支出调整后的实际可动用现金。

**计算公式**：
```
FreeCashFlow = NetIncome + Depreciation - CapitalExpenditure
```

> **说明**：当前模型折旧和资本支出均为 0，因此自由现金流等于净利润。

**业务含义**：企业实际可用于再投资、还债或分配的真实现金。

**依赖参数**：月净利润、折旧、资本支出


### 3.4 效率类指标定义

#### 每线索成本（Cost Per Lead, CPL）

**定义**：获取一个询盘线索的平均成本，以美元计价。

**计算公式**：
```
CPL = CPC / (ConversionRate × AllocationPercentage)
```

**业务含义**：衡量广告渠道的获客效率。CPL 越低，广告投入产出比越高。

**行业基准**：
- 全行业平均：\$70.11
- B2B 制造/工业：\$64.69–\$104.03
- 餐饮行业（Restaurants & Food）：\$30.27

**依赖参数**：CPC、CVR、分配比例


#### 客户获取成本（Customer Acquisition Cost, CAC）

**定义**：获取一个成交客户的总成本，以人民币计价。

**计算公式**：
```
CAC = AdvertisingExpense / Orders
```

> **说明**：当前模型销售佣金率为 0，因此 CAC 简化为广告费除以订单量。

**业务含义**：衡量销售与广告的综合获客效率。CAC 应低于客户生命周期价值（LTV），当前模型暂未计算 LTV。

**依赖参数**：广告费、月订单量


#### 广告支出回报率（Return on Ad Spend, ROAS）

**定义**：每投入 1 元广告费所获得的收入。

**计算公式**：
```
ROAS = Revenue / AdvertisingExpense
```

**业务含义**：衡量广告投放的直接经济效益。ROAS = 400% 意味着每 1 元广告费带来 4 元收入。

**行业基准**：通常认为 ROAS ≥ 400% 为健康水平。

**依赖参数**：月收入、广告费


### 3.5 平衡类指标定义

#### 盈亏平衡时间（Break-Even Time）

**定义**：从启动开始，首次实现单月净利润 ≥ 0 的月份。

**计算公式**：
```
盈亏平衡时间 = min{ t | 第 t 月净利润 ≥ 0 }
```

**业务含义**：回答“多久能开始赚钱”，是评估业务可行性的关键指标。

**依赖参数**：各月净利润（由模型计算）


#### 回本周期（Payback Period）

**定义**：从启动开始，累计净利润首次覆盖一次性启动成本的月份。

**计算公式**：
```
回本周期 = min{ t | Σ(第1月到第t月净利润) ≥ SetupInvestment }
```

**业务含义**：回答“多久能收回初始投入”。比盈亏平衡时间更严格，因为需要覆盖前期投入成本。

**依赖参数**：各月净利润、一次性启动成本


#### 6个月投资回报率（6-Month ROI）

**定义**：6 个月运营周期内，每投入 1 元启动资金所产生的净利润。

**计算公式**：
```
ROI = 6个月累计净利润 / SetupInvestment
```

**计算逻辑**：
- **分子（收益）**：6 个月累计净利润（已扣除所有运营成本：广告费、COGS、费用、税金等）
- **分母（投资）**：一次性启动成本（开模费、网站搭建、公司注册、样品费等）

**业务含义**：ROI = 5 倍意味着每投入 1 元启动资金，6 个月内赚回 5 元净利润。

**区分说明**：广告费、采购成本、月度运营费用被视为**运营成本**而非投资本金。这些成本已在净利润计算中扣除，因此不重复计入分母。这种分类符合创业项目财务分析的典型做法。

**依赖参数**：6 个月累计净利润、一次性启动成本

#### 8.2.1 累计净利润的区间估计

为评估 6 个月累计净利润的不确定性，本模型从三阶段各自的模拟结果中独立随机抽取并相加（独立叠加法），重复 5,000 次，得到 6 个月累计净利润的分布。该方法与第13章蒙特卡洛模拟方法一致，确保结论可比。

> **方法说明**：本处按三阶段独立随机采样后加总（非对齐加总），与第13章方法一致，确保结论可比。在独立同分布条件下，三阶段随机样本之间不存在内在关联，独立叠加法是估计总利润分布的合理方法。

**模拟结果**：

基于三阶段模拟结果的独立叠加法，6 个月累计净利润的 25 分位至 75 分位区间为：

| 统计量 | 6 个月累计净利润 |
|:---|:---|
| 25 分位 | ¥101,110 |
| 75 分位 | ¥199,608 |
| **中位** | **¥150,359** |

````python
# 1. Convert lists to lookup dictionaries (O(n))
exp_map = {s[vn.SYSTEM_RUN_ID]: s for s in income_simulation_experimental}
opt_map = {s[vn.SYSTEM_RUN_ID]: s for s in income_simulation_optimization}
sta_map = {s[vn.SYSTEM_RUN_ID]: s for s in income_simulation_stabilized}

aggregated_income_simulation = []

# 2. Iterate once (O(n))
for target_id in exp_map:
    # Dictionary lookups are O(1)
    exp = exp_map[target_id]
    opt = opt_map[target_id]
    sta = sta_map[target_id]

    aggregated_income_simulation.append({
        vn.SYSTEM_RUN_ID: target_id,
        vn.GROSS_PROFIT: 2 * (exp[vn.GROSS_PROFIT] + opt[vn.GROSS_PROFIT] + sta[vn.GROSS_PROFIT]),
        vn.NET_INCOME: 2 * (exp[vn.NET_INCOME] + opt[vn.NET_INCOME] + sta[vn.NET_INCOME])
    })

# print(aggregated_income_simulation)

aggregated_gross_profit_simulations = [simulation[vn.GROSS_PROFIT] for simulation in aggregated_income_simulation]
aggregated_net_income_simulations = [simulation[vn.NET_INCOME] for simulation in aggregated_income_simulation]

aggregated_income_distribution_functions = [
    lambda ax: generate_histogram_from_array(aggregated_income_simulation, vn.GROSS_PROFIT, ax=ax,
                                             title="Gross Profit in 6 Months"),
    lambda ax: generate_histogram_from_array(aggregated_income_simulation, vn.NET_INCOME,
                                             title="Net Income in 6 months", ax=ax),
]

fig = plot_multiple_views(aggregated_income_distribution_functions)
````

## 第14章：线性回归分析

### 14.1 方法说明

本章对每个参数分别建立一元线性回归模型，以净利润为因变量（Y），各参数为自变量（X），评估其对净利润变异的解释程度。

**指标解读**：

| 指标 | 含义 | 在本分析中的应用 |
|:---|:---|:---|
| **斜率** | X 变化 1 单位时 Y 的平均变化 | 反映变量间的联合分布关系，需结合弹性系数（第10章）综合解读 |
| **R²** | X 单独解释 Y 变异的比例（0-1，越高越好）| 比较各变量对净利润变异的相对解释力 |
| **p 值** | 统计显著性（< 0.05 为显著）| 排除随机性 |

> **注意**：
> - 本回归中所有变量同步随机采样，斜率反映的是变量间的联合分布关系，而非严格的因果边际贡献。因果解读请参见第10章弹性系数。
> - 斜率绝对值受量纲影响（如 CPC 单位为 \$、CVR 为百分点，而净利润为 ¥），解读时应以方向（正/负）和 R² 为主，斜率仅供辅助参考。


### 14.2 运行结果

基于 5,000 次模拟的线性回归结果如下：

| X 变量 | 斜率 | R² | p 值 | 弹性（第10章）|
|:---|:---|:---|:---|:---|
| Unit FOB Price | 17.64 | **17.01%** | < 0.001 | +1.34 |
| Unit EXW Price | -3.14 | **1.54%** | < 0.001 | -0.23 |
| Marketing Expense | 17.97 | **28.07%** | < 0.001 | +1.06 |
| CPC | -27,336.56 | 10.31% | < 0.001 | -1.00 |
| CVR | 1,166,212.03 | 15.95% | < 0.001 | +1.11 |
| Close Rate | 280,242.36 | 17.56% | < 0.001 | +1.11 |

> **关于斜率解读**：CVR 和 Close Rate 的斜率数值较大，原因是自变量单位为百分点（0.01），而净利润单位为元。为便于理解，可将斜率乘以 0.01 解读为“每提高 1 个百分点带来的净利润变化”：CVR 每提高 1 个百分点，净利润平均增加约 ¥11,662；Close Rate 每提高 1 个百分点，净利润平均增加约 ¥2,802。

### 15.2 核心风险提示

#### 15.2.1 模型假设风险

| 风险 | 说明 | 影响 | 缓解措施 |
|:---|:---|:---|:---|
| **Unit FOB Price 与 Close Rate 独立假设** | 模型假设提价不影响客户成交意愿。实际中，提价可能导致成交率下降 | 高估提价的利润效应 | 小幅度逐步提价（如5-10%），监控成交率变化 |
| **广告预算与效率独立假设** | 模型假设预算增加不影响CPC和CVR。实际中，大规模增预算可能导致CPC上升、CVR下降 | 高估大规模增预算的边际回报 | 分阶段增加预算，每阶段监控效率指标 |
| **线性叠加假设** | 模型假设各参数独立线性影响利润。实际中可能存在非线性交互效应 | 多变量同时优化时，实际效果可能低于线性叠加预期 | 一次只优化1-2个变量，验证后再扩展 |

#### 15.2.4 风险优先级总结

| 优先级 | 风险类别 | 核心风险 | 建议关注程度 |

### 附录D：模型局限性详细说明

本附录对第16章中概述的模型局限性进行详细展开，说明各项简化的具体影响及其对结论的可信度影响。

#### D.1 核心假设的详细影响分析

| 假设 | 详细说明 | 对结论的影响幅度 | 缓解措施 |
|:---|:---|:---|:---|
| **Unit FOB Price 与 Close Rate 独立** | 模型假设提价不影响客户成交意愿。实际中，价格提升可能抑制B端客户采购决策 | 售价弹性可能被高估 | 小幅度逐步提价（5-10%），监控Close Rate变化 |
| **广告模型线性** | 模型假设预算增加不改变CPC和CVR。实际中，预算扩大可能需投放竞争更激烈的关键词 | 高估大规模增预算的边际回报 | 分阶段增加预算，每阶段监控效率指标 |
| **无账期/现金流时序** | 模型假设收入当月确认，无付款延迟 | 前期资金需求被低估 | 准备3个月运营资金缓冲（约¥25,000-30,000）|
| **无供应商定金** | 模型假设采购成本全额计入当月COGS | 前期资金需求被低估 | 同上 |
| **物流成本为0** | 模型假设运费由下游买方承担 | 利润可能高估约5-10% | V2版本纳入物流成本 |
| **无升级包** | 模型忽略温控、影音等选配模块 | 收入可能低估10-15% | V2版本纳入升级包模块 |
| **无复购** | 模型假设每客户仅一次交易 | 长期价值低估 | V3版本引入LTV模型 |
| **单一广告渠道** | 模型仅追踪Google Search | 线索量可能低估 | V2版本纳入多渠道 |

#### D.2 结论可信度的详细评估

| 结论 | 可信度 | 详细说明 |
|:---|:---|:---|
| 盈利能力（净利润）| **高** | 核心成本结构（COGS、运营费用、广告费）已覆盖；多维度分析（弹性、R²、盈亏平衡）一致验证 |
| 盈亏平衡时间 | **高** | 第1个月盈利，安全边际充足；即使实际数据与模型有偏差，结论方向不会改变 |
| 回本周期 | **高** | 启动成本低（¥6,000-15,000），3个月内收回的结论稳健 |
| Unit FOB Price 是最优杠杆 | **中** | 需在实际运营中验证价格-成交率关系；V2需求曲线将修正这一结论 |
