#### 2.3 新价格模型框架

已废弃原 `Selling Price` 直接赋值的方式，改用以下公式：

```
FOB = Retail Price × (1 - Deduction Rate)

其中：
Deduction Rate = Channel Markup Rate + Ocean Freight Rate + Tariff Rate (占Retail Price)
```

**参数定义**：

| 参数 | 定义 | 悲观 | 基准 | 乐观 |
|:---|:---|:---|:---|:---|
| Retail Price | 北美终端零售价（美元/套）| \$5,000 | \$8,000 | \$12,000 |
| Channel Markup Rate | 下游渠道利润（占Retail Price）| 30% | 20% | 15% |
| Ocean Freight Rate | 海运费（含清关+本地物流，占Retail Price）| 15% | 8% | 4% |
| Tariff Rate（占Retail Price）| 关税（占Retail Price，基准假设FOB/Retail=55%）| 16.5% | 13.8% | 5.5% |
| **Deduction Rate** | 总扣除比例 | **61.5%** | **41.8%** | **24.5%** |
| **FOB** | 我们的出口报价（美元/套）| **\$3,080** | **\$4,656** | **\$6,040** |

**数据来源与采信度**：
- Retail Price：Gaxdome 品牌官网及行业价格指南，采信度中等
- Channel Markup Rate：建材/户外设备行业惯例，采信度中等
- Ocean Freight Rate：40尺柜运价折算，采信度较高
- Tariff Rate：美国301关税制度，采信度较高（HS编码待确认）

### 二、论证方法


#### 2.1 论证框架

本调研采用**三层交叉验证法**，确保结论的稳健性：

```
┌─────────────────────────────────────────────────────────────┐
│                    第一层：成本拆解法                         │
│        从原材料价格出发，逐项加总估算成品合理价格区间            │
│         PC板材成本 + 框架成本 + 加工费 + 工厂利润              │
└─────────────────────────────────────────────────────────────┘
                              ↓
┌─────────────────────────────────────────────────────────────┐
│                    第二层：市场报价法                         │
│        收集多个来源的公开报价，取合理区间作为交叉验证            │
│         行业文章报价 + B2B平台信息 + 厂家技术文档               │
└─────────────────────────────────────────────────────────────┘
                              ↓
┌─────────────────────────────────────────────────────────────┐
│                    第三层：逻辑一致性检验                      │
│        验证"采购价→售价→毛利率"是否处于行业合理范围             │
│             毛利率应处于30-70%区间（B2B贸易行业）              │
└─────────────────────────────────────────────────────────────┘
```


#### 2.2 第一层：成本拆解法（自下而上）

**核心逻辑**：一套完整PC星空房的出厂价 = 材料成本 + 加工费 + 工厂利润

**步骤一：板材成本估算**

| 等级 | 厚度 | 单价（元/㎡）| 单套用量（约25-35㎡）| 单套板材成本 |
|:---|:---|:---|:---|:---|
| 经济型 | 4mm | 35-50 | 30㎡ | ¥1,050-1,500 |
| 标准型 | 6mm | 50-80 | 30㎡ | ¥1,500-2,400 |
| 高端型 | 8-10mm | 100-150 | 30㎡ | ¥3,000-4,500 |

> **数据来源**：2025年Q3行业监测数据显示，国内PC耐力板主流价格区间为35-95元/㎡，8mm实心耐力板约68元/㎡，10mm约260元/㎡。价格差异主要来源于厚度、UV涂层、新料/回料比例。

**成本拆解法结论**：一套完整星空房的合理出厂价区间约为 **¥3,000-9,000**，这与财务模型中的乐观-悲观区间基本吻合。¥1,800的采购价仅能覆盖经济型配置的硬件成本，无法包含工厂合理利润。

| **悲观** | \$5,000 | 3-3.5m 小型款，基础配置 | 行业低端配置上限 |
| **基准** | \$8,000 | 4m 标准商用款，UV涂层+基础配件 | Gaxdome 基础款中位数 |
| **乐观** | \$12,000 | 4-5m 款，含温控/影音升级 | Gaxdome 中端款下限 |

---

### 三、Trade Discount Rate：折扣率拆解

`Trade Discount Rate` 由以下三个子项目组成：

```
Trade Discount Rate = 1 - (Channel Markup Rate + Ocean Freight Rate + Tariff Rate)
```

#### 3.1 下游渠道利润（Channel Markup Rate）

**数据来源**：建材/户外设备行业渠道惯例

| 渠道类型 | 加价率（占零售价）|
|:---|:---|
| B2B 工业设备分销 | 15-30% |
| 户外装备/建材零售 | 25-45% |

**建议取值**：

| 情景 | 取值 | 依据 |
|:---|:---|:---|
| 悲观 | 30% | 多层分销，下游强势 |
| 基准 | 20% | 单层分销，正常模式 |
| 乐观 | 15% | DTC 直销，无中间商 |

#### 3.2 海运费率（Ocean Freight Rate）

**数据来源**：
- 40尺柜运价：美西 \$3,000-6,000，美东 \$5,000-8,000
- 单柜装载量：10-15 套
- 折算单套运费：\$400-800

**建议取值**（占 Retail Price 的比例）：

| 情景 | 取值 | 依据 |
|:---|:---|:---|
| 悲观 | 15% | 旺季运价高 + 零售价偏低 |
| 基准 | 8% | 行业平均水平 |
| 乐观 | 4% | 淡季运价低 + 零售价偏高 |

#### 3.3 关税率（Tariff Rate）

**数据来源**：
- 美国 301 关税制度（2018 年起对中国商品加征）
- PC 星空房可能归入 HS 编码 3926.90（塑料制品）或 9406.90（预制建筑）
- 基础关税 2.5%-5.5% + 301 加征 25%

**建议取值**（占 FOB 的比例）：

| 情景 | 取值 | 依据 |
|:---|:---|:---|
| 悲观 | 30% | 高基础关税 + 301 关税全额征收 |
| 基准 | 25% | 301 关税 25% 全额征收 |
| 乐观 | 10% | 获得部分关税豁免 |

> 注：关税占 Retail Price 的比例需乘以 FOB/Retail Price 系数。基准情景下（假设 FOB/Retail Price = 55%），关税占 Retail Price 约 13.8%。

---

### 四、Trade Discount Rate 汇总

| 情景 | 下游利润 | + 海运费 | + 关税 | = 总扣除 | **Trade Discount Rate** |
|:---|:---|:---|:---|:---|:---|
| 悲观 | 30% | 15% | 16.5% | 61.5% | **38.5%** |
| 基准 | 20% | 8% | 13.8% | 41.8% | **58.2%** |
| 乐观 | 15% | 4% | 5.5% | 24.5% | **75.5%** |

> 关税占 Retail Price 的比例按 FOB/Retail Price = 55% 折算，模型实际计算时应采用动态公式。

---

### 五、Trade Price 最终取值

以 Retail Price 基准值 \$8,000 为例：

| 情景 | Retail Price | × Trade Discount Rate | = **Trade Price (FOB)** |
|:---|:---|:---|:---|
| 悲观 | \$8,000 | 38.5% | **\$3,080** |
| 基准 | \$8,000 | 58.2% | **\$4,656** |
| 乐观 | \$8,000 | 75.5% | **\$6,040** |

---

### 六、数据有效性总结

| 参数 | 数据充分度 | 主要局限性 |
|:---|:---|:---|
| Retail Price | ★★★☆☆ | 仅一个品牌公开数据 |
| 下游渠道利润 | ★★★☆☆ | 基于相似产品推算 |
| 海运费率 | ★★★★☆ | 运价波动大 |
| 关税率 | ★★★★☆ | 政策变化频繁，HS编码待确认 |

**建议**：在正式采购/销售前，通过实际询价和海关归类咨询验证上述参数。
